```python
from pymatgen.core import Composition

transition_metals = ['Ti','V','Cr','Mn','Fe','Co','Ni','Cu','Mo','Nb','W','Ru']

def get_transition_metal(formula):
    comp = Composition(formula)
    tms = [el.symbol for el in comp.elements if el.symbol in transition_metals]
    return '-'.join(sorted(tms)) if tms else 'None'

df['tm_group'] = df['formula_discharge'].apply(get_transition_metal)

df['tm_group'].head()
```
