```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Final Research Project - Sheet20.csv')

cols_to_analyze = ['average_voltage', 'capacity_grav', 'energy_grav',
                   'max_delta_volume', 'fracA_charge', 'fracA_discharge',
                   'stability_charge', 'stability_discharge']
numeric_df = df[cols_to_analyze]
correlation_matrix = numeric_df.corr(method="spearman")

# In a .py script you must use print() — unlike Jupyter, nothing shows automatically
print(correlation_matrix)
print(f"\nData Size: {len(df)} rows")

plt.figure(figsize=(11, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Optimal Battery Properties - Optimal Dataset")
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig("correlation_heatmap.png", dpi=150)
print("\nHeatmap saved to correlation_heatmap.png")
plt.show()
```
